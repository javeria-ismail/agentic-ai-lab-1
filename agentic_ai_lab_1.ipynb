```python
def weather_advice(temperature_C, humidity_percent, windspeed, pressure):
    advice = []

    # Temperature
    if temperature_C >= 35:
        advice.append("Extreme heat!")
    elif temperature_C <= 5:
        advice.append("Cold weather!")

    # Humidity
    if humidity_percent >= 80:
        advice.append("High humidity!")

    # Wind Speed
    if windspeed >= 50:
        advice.append("Storm warning!")
    elif windspeed >= 39:
        advice.append("Gale force winds!")
    elif windspeed >= 29:
        advice.append("Strong winds!")
    elif windspeed >= 12:
        advice.append("Moderate breeze!")
    else:
        advice.append("Calm winds!")

    # Atmospheric Pressure (new rule)
    if pressure < 1000:
        advice.append("Low pressure: Rain or storm possible!")
    elif pressure > 1025:
        advice.append("High pressure: Clear weather expected!")
    else:
        advice.append("Normal atmospheric pressure!")

    if not advice:
        advice.append("Conditions look moderate.")

    return advice


if __name__ == "__main__":

    sample_days = [
    (45, 98, 65, 985),
    (5, 30, 6, 1028),
    (19, 80, 30, 1012),
    (37, 70, 14, 999),
    (25, 50, 11, 1022)
    ]

    for temp, humidity, windspeed, pressure in sample_days:
        result = weather_advice(temp, humidity, windspeed, pressure)

        print(
            f"Temp={temp}°C, "
            f"Humidity={humidity}%, "
            f"Wind={windspeed} km/h, "
            f"Pressure={pressure} hPa"
        )

        for item in result:
            print("-", item)

        print()
```

```python
# Task 2 - Simple Data-Driven Spam Classifier

spam_keywords = [
    "copy",
    "loss",
    "lazy",
    "no click ",
    "prey"
]

threshold = 2  


def classify_email(email_text):
    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= threshold:
        return "Spam", count
    else:
        return "Legitimate", count


# Six test emails
test_emails = [
    {
        "text": "Congratulations! You loss. Click here to claim your prize.",
        "actual": "Spam"
    },
    {
        "text": "Urgent notice! no click  to receive your free gift.",
        "actual": "Spam"
    },
    {
        "text": "Meeting tomorrow at 10 AM. not be lazy  Please review the attached document.",
        "actual": "Legitimate"
    },
    {
        "text": "You have won a free prize. prey is a word!",
        "actual": "Spam"
    },
    {
        "text": "Reminder: Your library book is due next week.",
        "actual": "Legitimate"
    },
    {
        "text": "Urgent update regarding your account settings.",
        "actual": "Legitimate"
    }
]

correct = 0

print("===== SPAM CLASSIFIER RESULTS =====\n")

for i, email in enumerate(test_emails, start=1):
    prediction, keyword_count = classify_email(email["text"])

    if prediction == email["actual"]:
        correct += 1

    print(f"Email {i}")
    print(f"Text: {email['text']}")
    print(f"Keywords Found: {keyword_count}")
    print(f"Predicted: {prediction}")
    print(f"Actual: {email['actual']}")
    print("-" * 50)

accuracy = (correct / len(test_emails)) * 100

print(f"\nAccuracy: {accuracy:.2f}%")
```

```python
# Task 3 - Simple Data-Driven Spam Classifier

spam_keywords = [
    "copy",
    "loss",
    "lazy",
    "no click",
    "prey"
]

threshold = 1


# Like facial recognition and medical image diagnosis, this classifier assigns an input 
#to one of a fixed set of categories.
def classify_email(email_text):
    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= threshold:
        return "Spam", count
    else:
        return "Legitimate", count


# Six test emails
test_emails = [
    {
        "text": "You have suffered a loss. Copy this message to claim your reward.",
        "actual": "Spam"
    },
    {
        "text": "No click this link. Copy the message and send it to your friends.",
        "actual": "Spam"
    },
    {
        "text": "Meeting tomorrow at 10 AM. Please do not be lazy. Review the attached document.",
        "actual": "Legitimate"
    },
    {
        "text": "This is a warning about your loss. Do not prey on others with this message.",
        "actual": "Spam"
    },
    {
        "text": "Reminder: Your library book is due next week.",
        "actual": "Legitimate"
    },
    {
        "text": "Urgent update regarding your account settings.",
        "actual": "Legitimate"
    }
]


correct = 0

print("===== SPAM CLASSIFIER RESULTS =====\n")


for i, email in enumerate(test_emails, start=1):

    prediction, keyword_count = classify_email(email["text"])

    if prediction == email["actual"]:
        correct += 1

    print(f"Email {i}")
    print(f"Text: {email['text']}")
    print(f"Keywords Found: {keyword_count}")
    print(f"Predicted: {prediction}")
    print(f"Actual: {email['actual']}")
    print("-" * 50)


accuracy = (correct / len(test_emails)) * 100

print(f"\nAccuracy: {accuracy:.2f}%")
```
